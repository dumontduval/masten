```python
import pandas as pd
from pathlib import Path
import matplotlib.pyplot as plt 
import numpy as np
```

```python
path_wd = Path.cwd()
path_dir = path_wd.parent
```

```python
df_carbon = pd.read_csv(path_dir / 'data/df_fuel_ckan (4).csv', index_col=0)
df_carbon.index = pd.to_datetime(df_carbon.index)
df_carbon
```

```python
df_carbon
```

```python
plt.plot(df_carbon['CARBON_INTENSITY'])
plt.ylabel('gCO₂/kWh')
plt.title('Carbon intensity over the years')
```

```python
data = df_carbon['CARBON_INTENSITY'].loc['2024':'2025']

threshold = data.quantile(0.75)
colors = np.where(data > threshold, 'red', 'blue')

plt.scatter(np.arange(len(data)), data, c=colors, s=1)
plt.axhline(threshold, color='gray', linestyle='--')
plt.ylabel('gCO₂/kWh')
plt.title('Carbon intensity over the years 2024-2025')
plt.show()
```

```python
data = df_carbon['CARBON_INTENSITY'].loc['2024']

# Calculate 75th percentile
q75 = data.quantile(0.75)

# Plot the series
plt.figure(figsize=(10, 5))
plt.plot(data, label='Carbon Intensity')

# Add 75th percentile line
plt.axhline(q75, color='red', linestyle='--', label=f'75th Percentile ({q75:.1f})')

# Add legend and labels
plt.title('Carbon Intensity with 75th Percentile Line')
plt.xlabel('Date')
plt.ylabel('gCO₂/kWh')
plt.legend()
plt.show()
```
